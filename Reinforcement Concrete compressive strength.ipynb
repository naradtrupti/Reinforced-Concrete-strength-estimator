```python
#make your own class
class  MeraLR:
    def __init__(self):
         self.coef_ = None
         self.intercept_ = None

    def fit(self,X_train,y_train):
        X_train = np.insert(X_train,0,1,axis=1)

        #Calculate the coeffs
        betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
        self.intercept_=betas[0]
        self.coef_=betas[1:]

    def predict(self,X_test):
        y_pred = np.dot(X_test,self.coef_) + self.intercept_
        return y_pred
```

```python
#APPLY GRADIENT DECENT JUST FOR LEARNING PURPOSE

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_selected)
x_test_scaled = scaler.transform(X_test_selected)
```

```python
import numpy as np 

n_samples,n_features = X_train_scaled.shape

weights = np.zeros(n_features)
bias = 0

learning_rate = 0.01 
epochs = 1000

print("Initial weights:", weights)
print("Initial bias:", bias)
```

```python
m = len(y_train)

cost_history = []

for i in range(epochs):

    # Predictions
    y_pred_gd = np.dot(X_train_scaled, weights) + bias

    # Calculate errors
    error = y_pred_gd - y_train.to_numpy()

    # Calculate gradients
    dw = (2/m) * np.dot(X_train_scaled.T, error)
    db = (2/m) * np.sum(error)

    # Update weights and bias
    weights = weights - learning_rate * dw
    bias = bias - learning_rate * db

    # Calculate cost (MSE)
    cost = np.mean(error**2)
    cost_history.append(cost)

print("Final Weights:", weights)
print("Final Bias:", bias)
print("Final Cost:", cost_history[-1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(cost_history)

plt.xlabel("Iterations (Epochs)")
plt.ylabel("Mean Squared Error (Cost)")
plt.title("Cost Reduction using Gradient Descent")

plt.show()
```

```python
#Predictions making using gradient decent
y_pred_gd = np.dot(x_test_scaled,weights) + bias

print("First 10 Predicted Values:")
print(y_pred_gd[:10])
```

```python
df_fe = df.copy()
binder_cols = ['cementecontent', 'silicafumecontent', 'slagcontent',
               'flyashcontent', 'metakaolincontent']
df_fe['binder'] = df_fe[binder_cols].sum(axis=1)
df_fe['w_b'] = df_fe['water'] / df_fe['binder']

base12 = list(X.columns)
fe_cols = base12 + ['binder', 'w_b']
```

```python
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.ensemble import GradientBoostingRegressor

# Rows with the same base mix (only fiber differs) share a group
mix_cols = ['fine aggregate', 'coarse aggregate', 'water', 'additivecontent',
            'silicafumecontent', 'slagcontent', 'flyashcontent',
            'metakaolincontent', 'cementecontent']
groups = pd.Series(
    df_fe[mix_cols].round(2).astype(str).agg('|'.join, axis=1).factorize()[0],
    index=df_fe.index
)

rf = lambda: RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)
candidates = {
    "RF  6 selected":   (rf(), selected_features),
    "RF  12 features":  (rf(), base12),
    "RF  12 + w_b":     (rf(), fe_cols),
    "GBR 12 + w_b":     (GradientBoostingRegressor(random_state=42), fe_cols),
}

gkf = GroupKFold(n_splits=5)
for name, (model_, cols) in candidates.items():
    s = cross_val_score(model_, df_fe.loc[X_train.index, cols], y_train,
                        groups=groups.loc[X_train.index], cv=gkf, scoring="r2")
    print(f"{name:18s} mean R2 = {s.mean():.4f}  std = {s.std():.4f}")
```

```python
from sklearn.model_selection import GroupShuffleSplit

best_cols = selected_features
make_model = lambda: RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)

# Original random test split (optimistic)
m1 = make_model().fit(df.loc[X_train.index, best_cols], y_train)
p1 = m1.predict(df.loc[X_test.index, best_cols])
print("Random split  R2:", round(r2_score(y_test, p1), 4),
      " MAE:", round(mean_absolute_error(y_test, p1), 3))

# Honest split: test mixes never seen in training
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
tr, te = next(gss.split(df, groups=groups))
m2 = make_model().fit(df.iloc[tr][best_cols], df.iloc[tr]['compressivestrength'])
p2 = m2.predict(df.iloc[te][best_cols])
print("Unseen-mix    R2:", round(r2_score(df.iloc[te]['compressivestrength'], p2), 4),
      " MAE:", round(mean_absolute_error(df.iloc[te]['compressivestrength'], p2), 3))
```

```python
loaded_model = joblib.load("concrete_final_model.pkl")
meta = joblib.load("concrete_final_meta.pkl")

def predict_concrete_strength(inputs: dict):
    missing = [c for c in meta["features"] if c not in inputs]
    if missing:
        raise ValueError(f"Missing inputs: {missing}")

    row = pd.DataFrame([inputs])[meta["features"]]

    outside = [c for c in meta["features"]
               if not (meta["min"][c] <= row[c].iloc[0] <= meta["max"][c])]
    if outside:
        print("Warning: outside training range for:", outside)

    return float(loaded_model.predict(row)[0])
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib

st.set_page_config(page_title="Concrete Strength Estimator", page_icon="🏗️")

@st.cache_resource
def load_files():
    return joblib.load("concrete_final_model.pkl"), joblib.load("concrete_final_meta.pkl")

model, meta = load_files()
features = meta["features"]

labels = {
    "silicafumecontent": "Silica fume",
    "additivecontent": "Additive (admixture)",
    "water": "Water",
    "cementecontent": "Cement",
    "flyashcontent": "Fly ash",
    "fibercontent": "Fiber content",
}

st.title("🏗️ Fiber-Reinforced Concrete Strength Estimator")
st.caption("Predicts compressive strength (MPa) from mix proportions. "
           "Random Forest trained on 521 literature mixes.")

st.subheader("Mix proportions")
inputs = {}
cols = st.columns(2)
for i, f in enumerate(features):
    with cols[i % 2]:
        inputs[f] = st.number_input(
            labels.get(f, f),
            min_value=0.0,
            value=float(meta["median"][f]),
            step=1.0,
            help=f"Training range: {meta['min'][f]:.1f} to {meta['max'][f]:.1f}",
        )

if st.button("Predict strength", type="primary"):
    row = pd.DataFrame([inputs])[features]

    outside = [labels.get(c, c) for c in features
               if not (meta["min"][c] <= inputs[c] <= meta["max"][c])]
    if outside:
        st.warning("Outside the training range for: " + ", ".join(outside)
                   + ". The prediction may be unreliable.")

    per_tree = np.array([t.predict(row.values)[0] for t in model.estimators_])
    mean_pred, spread = float(per_tree.mean()), float(per_tree.std())

    st.metric("Estimated compressive strength", f"{mean_pred:.1f} MPa")
    st.write(f"Tree disagreement: ±{spread:.1f} MPa")
    st.write("Typical error on new, unseen mixes is about **±8 MPa**, "
             f"so a realistic range is roughly **{mean_pred - 8:.0f} to {mean_pred + 8:.0f} MPa**.")

st.info("This is an estimate from 521 literature mixes and does not include curing age. "
        "Verify with lab testing. Not for structural design use.")
```
